```python
import pandas as pd
import numpy as np
from scipy.interpolate import interp1d

# 1. LOAD THE DATA
# We skip the first 6 lines of metadata to get to the header
try:
    df = pd.read_csv('../data/ECG/ECGPeakTable.csv', skiprows=6)
except:
    # Fallback if the file was edited and metadata removed
    df = pd.read_csv('../data/ECG/ECGPeakTable.csv')

# 2. FILTERING
# We only want the movie segments. 
# It appears labeled as "PC" for some and "PCM" for others.
movie_labels = ['PC']#, 'PCM']
df_movie = df[df['Stimulus Label'].isin(movie_labels)].copy()

# Remove rows where IBIs are NA (usually the first beat of a segment)
df_movie = df_movie.dropna(subset=['IBIs'])

# 3. CALCULATE INSTANTANEOUS HR & ALIGNED TIME
# HR = 60,000 / IBI (since IBI is in ms)
df_movie['HeartRate'] = 60000 / df_movie['IBIs']

# Calculate time relative to the start of the stimulus in Seconds
# (Peak Time - Stimulus Start) / 1000
df_movie['Time_Rel_Sec'] = (df_movie['Peak Time'] - df_movie['Stimulus Start']) / 1000

# 4. RESAMPLING / INTERPOLATION
# We need a common time vector for everyone.
# Let's assume the movie is roughly 320 seconds long (approx 5.3 mins based on your data).
# We will sample at 1Hz (1 reading per second). You can change '1' to '0.5' for 2Hz, etc.
sampling_rate_hz = 1 
common_time_grid = np.arange(0, 320, 1/sampling_rate_hz)

aligned_data = {'Time (s)': common_time_grid}

# Get list of unique participants
participants = df_movie['Respondent Name'].unique()

for sub in participants:
    # Get data for this specific person
    sub_data = df_movie[df_movie['Respondent Name'] == sub].sort_values('Time_Rel_Sec')
    
    # We need strictly increasing time for interpolation
    # (Sometimes R-peak detection creates artifacts, dropping duplicates ensures stability)
    sub_data = sub_data.drop_duplicates(subset='Time_Rel_Sec')
    
    # Create the interpolation function
    # kind='linear' draws a straight line between beats. 
    # kind='cubic' makes it smoother but can overshoot if data is noisy.
    if len(sub_data) > 1:
        f = interp1d(sub_data['Time_Rel_Sec'], sub_data['HeartRate'], 
                     kind='linear', bounds_error=False, fill_value=np.nan)
        
        # Apply function to our common time grid
        interpolated_hr = f(common_time_grid)
        
        # Clean up the column name (e.g., "sub002")
        col_name = sub.split('_')[0] 
        aligned_data[col_name] = interpolated_hr

# 5. CREATE FINAL DATAFRAME
final_df = pd.DataFrame(aligned_data)

# Show the first few rows
print(final_df.head(10))

# Save to CSV for your ISC analysis
final_df.to_csv('../data/ECG/Processed_HeartRate_TimeSeries.csv', index=False)
```
