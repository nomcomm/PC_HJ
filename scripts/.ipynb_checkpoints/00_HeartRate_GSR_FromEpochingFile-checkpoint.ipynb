```python
import pandas as pd

# 1. LOAD
# Fix: Change header to 6 (Line 7 of the file)
df = pd.read_csv('../data/Epoch/Epoching.csv', header=6)

# Clean up column names (remove accidental spaces)
df.columns = df.columns.str.strip()

# 2. FILTER
# Keep only the movie segments
df = df[df['Label'].isin(['PC', 'PCM'])].copy()

# Clean subject names (turn "sub002_11-11-..." into "sub002")
df['Respondent Name'] = df['Respondent Name'].apply(lambda x: x.split('_')[0])

# ---------------------------------------------------------
# PROCESS HEART RATE (The first "Mean" column)
# ---------------------------------------------------------
# Select columns. Note: The first "Mean" is Heart Rate
df_hr = df[['Respondent Name', 'Epoch', 'Mean', 'Quality']].copy()
df_hr = df_hr.rename(columns={'Mean': 'HeartRate'})

# Pivot: col=Subject, row=Time
hr_matrix = df_hr.pivot(index='Epoch', columns='Respondent Name', values='HeartRate')
hr_matrix.reset_index(inplace=True)
hr_matrix['Epoch'] = hr_matrix['Epoch'] - 1 
hr_matrix = hr_matrix.rename(columns={'Epoch': 'Time (s)'})

# Save Heart Rate
hr_matrix.to_csv('../data/Epoch/HeartRate_Matrix.csv', index=False)
print("Heart Rate Matrix Created: HeartRate_Matrix.csv")

# ---------------------------------------------------------
# PROCESS GSR (The second "Mean" column, automatically named "Mean.1")
# ---------------------------------------------------------
# Select columns. Note: "Mean.1" corresponds to the GSR columns on the right side of the CSV
df_gsr = df[['Respondent Name', 'Epoch', 'Mean.1', 'Quality.1']].copy()
df_gsr = df_gsr.rename(columns={'Mean.1': 'GSR', 'Quality.1': 'GSR_Quality'})

# Pivot: col=Subject, row=Time
gsr_matrix = df_gsr.pivot(index='Epoch', columns='Respondent Name', values='GSR')
gsr_matrix.reset_index(inplace=True)
gsr_matrix['Epoch'] = gsr_matrix['Epoch'] - 1 
gsr_matrix = gsr_matrix.rename(columns={'Epoch': 'Time (s)'})

# Save GSR
gsr_matrix.to_csv('../data/Epoch/GSR_Matrix.csv', index=False)
print("GSR Matrix Created: GSR_Matrix.csv")

# Print preview of GSR data
print("\nPreview of GSR Data:")
print(gsr_matrix.head())
```
